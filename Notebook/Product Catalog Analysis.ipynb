```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv(r'C:\Users\Hp\Desktop\Python\Product Catalog Dataset\product_catalog.csv')
df.head()
```

```python
df.loc[np.argsort(df['unit_price'].to_numpy())[-10:][::-1], ['product_name', 'unit_price']]
```

```python
df['profit'] = df['unit_price'] - df['product_cost']
df.head()
```

```python
df.loc[np.argsort(df['profit'].to_numpy())[-10:][::-1],['product_name','unit_price','product_cost','profit']]
```

```python
price = df['unit_price'].to_numpy()
{'Mean': int(price.mean()),'Median': int(np.median(price)),'Minimum': int(price.min()),'Maximum': int(price.max()),'Standard Deviation': int(price.std())}
```

```python
df.groupby('product_category')['unit_price'].mean().sort_values(ascending = False).head(10)
```

```python
df['margin_percentage'] = ((df['unit_price'] - df['product_cost']) / df['unit_price']) * 100
df.groupby('product_name')['margin_percentage'].sum().sort_values(ascending = False).head(10)
```

```python
def above_avg(unit_price):
    return (unit_price > unit_price.mean()).sum()

int(above_avg(df['unit_price']))    
```

```python
df.groupby('brand')['profit'].sum().sort_values(ascending = False).head(10)
```

```python
rating = df['product_rating'].to_numpy()
price = df['unit_price'].to_numpy()

np.corrcoef(rating, price)[0, 1]
```
